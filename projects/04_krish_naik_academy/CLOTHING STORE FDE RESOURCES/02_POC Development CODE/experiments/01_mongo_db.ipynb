# MongoDB CRUD with PyMongo

Connects to a MongoDB Atlas cluster and walks through create, read, update and delete on an `ecom_db` database. Set `MONGO_URI` in your `.env` before running.

```python
from pymongo import MongoClient
from pymongo.server_api import ServerApi
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
uri = os.environ["MONGO_URI"]  # mongodb+srv://<user>:<password>@<cluster>/...

client = MongoClient(uri, server_api=ServerApi('1'))
# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
db = client["ecom_db"]

collection = db["orders"]

print("order collection inilitalized")
```

```python
order = {
    "name" : "Baggy Jeans",
    "price" : 999 , 
    "category" : ["Men" , "Women" , "Kids"]
}

order 
```

```python
type(order)
```

```python
collection.insert_one(order)
```

CRUD 

```python
collection = db["products"]
```

```python
products = [
    {"name": "Jeans", "price": 1499, "category": "men"},
    {"name": "Kurti", "price": 999, "category": "women"},
    {"name": "Kids Jacket", "price": 1299, "category": "kids"}
]

collection.insert_many(products)
```

FIND - read 

```python
all_products = collection.find()

for p in all_products:
    print(p)
```

TEXT 2 NOSQL

```python
product = collection.find_one({"name": "Kurti"})

print(product)
```

```python
men_products = collection.find({"category": "men"})

for p in men_products:
    print(p)
```

UPDATE 

```python
collection.update_one(
    {"name": "Jeans"},
    {"$set": {"price": 1399}}
)

print("Product Updated ✅")
```

DELETE

```python
collection.delete_one({"name": "Kids Jacket"})

print("Product Deleted ✅")
```

```python
collection.delete_many({})

print("All Products Deleted ✅")
```

```python
collection.drop()

print("Collection Dropped ✅")
```
